## Правило: когда использовать Restart / Run all

**Restart Kernel** — нужно, когда:
- Хочешь проверить, что notebook работает с нуля
- В памяти остались старые переменные, которые мешают
- Что-то странно работает, и ты хочешь начать заново

**Run All** — нужно, когда:
- Хочешь запустить все ячейки по порядку, сверху вниз
- Проверяешь, что весь notebook работает корректно
- Перед сдачей работы — чтобы убедиться, что всё работает

**Правило:** Если в notebook есть ошибки из-за порядка ячеек — нужно сделать Restart Kernel, а затем Run All. Тогда все ячейки выполнятся по порядку, и ошибок не будет.

**Retrieval quiz ЛР 1
LR1-RQ1. Что такое runtime?**
Runtime (среда выполнения / сеанс) — это временная среда, в которой выполняется код. Она хранит значения переменных, созданные во время работы, и существует, пока активен сеанс. После перезапуска runtime значения переменных исчезают, хотя сам файл notebook сохраняется.

**LR1-RQ2. Почему порядок выполнения важнее визуального расположения ячеек?**
Потому что состояние runtime зависит именно от порядка запуска ячеек, а не от того, как они расположены на странице. Если ячейки выполнялись в другом порядке, переменные могут быть не созданы к моменту запуска зависимой ячейки, что приведёт к ошибке (например, NameError). Визуальный порядок в документе не гарантирует, что код выполнялся именно так.

**LR1-RQ3. Какое действие проверяет воспроизводимость: скриншот или Run all?**
Воспроизводимость проверяет Run all после перезапуска runtime (Restart runtime → Run all). Скриншот не является доказательством воспроизводимости, так как не показывает, что notebook выполняется сверху вниз без ошибок в чистой среде.

**LR1-RQ4. Что должно быть в Markdown-ячейке рядом с экспериментом?**
В Markdown-ячейке рядом с экспериментом должно быть пояснение: что проверялось, какой получен результат, как его интерпретировать. Также могут быть описаны обычный и ошибочный тестовые случаи, выводы о состоянии runtime. Markdown делает работу понятной и проверяемой.

**LR1-RQ5. Что нельзя публиковать в notebook?**
Нельзя публиковать персональные данные, токены, пароли и другую конфиденциальную информацию. В файле прямо указано: «не вставлять персональные данные», «без персональных данных; без токенов». Данные должны быть безопасными для публикации.
